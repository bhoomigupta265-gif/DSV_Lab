#Tuple
##orderend collection of items ()
## one main difference with list is tuple is immutable
# var_name=list[[stu1:()],[stu2:()]....]

```python
#create a tuple
student=(101,"Sanjeev","CSE",8.9)
print(student)

#tuple using index
print("Student Information in first location:",student[0])
print(student[1])

#print negetive index
print("Example for negetive index: ",student[-2])
```

```python
#To find the length of the tuple
subjects=("DSA","DSV","OS","Python","Machine Learning")
print(len(subjects))
```

```python
#Sloicing
subjects=("DSV","DSA","Statistics","SQL","Machine Learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
#Tuple immutable
students_info=("Chirag",21,"CSE",8.7)
students_info[3]=9.1
print(students_info)
```

```python
#Nested tuple
student=(101,"Rahul",(89,78,66))
print(student)
print(student[2])
print(student[2][1])
```

```python
#count function
marks=(75,78,68,75,89,75,92,98)
print("Time of Occurance: ",marks.count(75))
```

```python
#find the element present in the tuple or not
subjects=("Python","DSV","SQL","Machine Learning","Statistics")
print("Python" in subjects)
print("Java" in subjects)
```

# Set
## unordered collection of items
##mutable
##allows the different type of data
## it doesn't support for index
##it gives output with uniques ( NO Duplications )


```python
#Symbol for set { }
#Create a set with similar items
marks ={75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
#create a empty set
a={}
a=set()
#check its type
print(type(a))
```

```python
#Set doesn't support the index
students={"Amit","Riya","Priya"}
print(students[1])
```

```python
# add an item in the set
students={"Amit","Kiran","Sneha"}
students.add("Priya")
print(students)
# to add multiple values or items
students.update(["Ram","Chitra","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Priya")
print(students)
students.discard("Amit")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.remove("Karan")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.discard("Karan")
print(students)
```

```python
#create a set and clear list of items
students_info={"Amit","Karan","Riya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99.88,77,66}
del marks
print(marks)
```

```python
# set operation
python_students={"Amit","Neha","Sanjay","Manish"}
java_students={"Amit","Karan","Dhilip","Manish","Neha"}

# for union operation by using set

result=python_students.union(java_students)
print(result)

# for intersection operation by using set

result_intersection=python_students.intersection(java_students)
print("Intersection Operation:",result_intersection)

# to find difference with two set
res_diff=python_students.difference(java_students)
print("For Difference: ",res_diff)
```

```python
# convert to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
uniq_roll_no=list(unique_stud_id)
print(uniq_roll_no)
```
